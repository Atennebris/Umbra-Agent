# Umbra Session Analysis

This notebook analyzes session data from the Umbra CLI daemon.

```python
import numpy as np
import pandas as pd
from pathlib import Path

DATA_DIR = Path('~/.umbra/sessions')
```

## Data Loading

Load session logs from the SQLite database.

```python
def load_sessions(db_path: str) -> pd.DataFrame:
    """Load all sessions from the umbra DB."""
    import sqlite3
    conn = sqlite3.connect(db_path)
    return pd.read_sql('SELECT * FROM sessions', conn)

def compute_token_stats(df: pd.DataFrame) -> dict:
    return {
        'mean': df['tokens'].mean(),
        'p95': df['tokens'].quantile(0.95),
    }
```

## Model Analysis

Analyze model usage patterns.

```python
class ModelUsageAnalyzer:
    def __init__(self, sessions: pd.DataFrame):
        self.sessions = sessions

    def by_model(self):
        return self.sessions.groupby('model')['tokens'].sum()
```
